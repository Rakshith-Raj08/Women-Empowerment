```python
import pandas as pd

# Update this path to wherever you extracted the .dta file
file_path = "IAIR7EFL.DTA"  # match the exact filename in your folder

# Verified column list — confirmed directly against this dataset's own
# .DO dictionary file, not from memory. See chat for the verification log.
needed_columns = [
    'caseid',   # unique respondent ID
    'd005',     # correct weight for this subsample (Domestic Violence module weight)
    'sdweight', # state-level DV module weight — use this for state-wise comparisons
    'v024',     # state
    'v025',     # urban/rural
    'v106',     # highest educational level
    'v511',     # age at first cohabitation (marriage)
    'v714',     # currently working
    'v170',     # has a bank/financial account she uses
    'v739',     # who decides how earnings are spent
    'v743a',    # who decides on respondent's healthcare
    'v743b',    # who decides on large household purchases
    'v743d',    # who decides on visits to family/relatives
    's929',     # has money of her own she alone decides how to use (India-specific)
]

chunks = []
reader = pd.read_stata(
    file_path,
    columns=needed_columns,
    chunksize=50000,
    convert_categoricals=False,  # keep raw numeric codes for now
)

for chunk in reader:
    chunks.append(chunk)

df = pd.concat(chunks, ignore_index=True)

print("Shape:", df.shape)  # expect ~724,115 rows, 13 columns
print(df.head())
print("\nMissing value counts:\n", df.isna().sum())

df.to_csv("nfhs5_empowerment_subset.csv", index=False)
print("\nSaved nfhs5_empowerment_subset.csv")
```

```python
import pandas as pd
df = pd.read_csv("nfhs5_empowerment_subset.csv")

# Keep only respondents who received both the status module (d005 weight present)
# AND the decision-making questions (v743a present, i.e. ever-married).
# Verified earlier: within this group, v024, v025, v106, v511, v714, v170,
# v743b, v743d, s929 are all 100% complete already.
clean = df[df['d005'].notna() & df['v743a'].notna()].copy()

# v739 ("who decides how earnings are spent") is a skip-pattern question --
# only asked of women who are currently earning (v714 == 1). Missing here
# means "not applicable / no personal earnings", NOT a data gap.
# 0 is unused in v739's real coding (1,2,4,5 are the actual answers),
# so it's a safe, non-conflicting marker for this case.
clean['v739'] = clean['v739'].fillna(0)

print("Final clean sample size:", len(clean))
print()
print("Sanity check -- missing values remaining per column:")
print(clean.isna().sum())
print()
print("State coverage (number of states/UTs represented):", clean['v024'].nunique())

clean.to_csv("nfhs5_empowerment_clean.csv", index=False)
print("\nSaved nfhs5_empowerment_clean.csv -- ready for PCA/clustering step.")
```

```python
import pandas as pd

df = pd.read_csv("nfhs5_empowerment_clean.csv")

# --- Decision-making variables: convert category codes into an autonomy score ---
# 1 = Respondent alone            -> 2 (high autonomy)
# 2 = Respondent + husband/partner -> 1 (medium/shared autonomy)
# 3 = Respondent + other person    -> 1 (medium/shared autonomy)
# 4 = Husband/partner alone        -> 0 (low autonomy)
# 5 = Someone else                 -> 0 (low autonomy)
# 6 = Other                        -> 0 (low autonomy)
autonomy_map = {1: 2, 2: 1, 3: 1, 4: 0, 5: 0, 6: 0}

df['healthcare_autonomy']  = df['v743a'].map(autonomy_map)
df['purchases_autonomy']   = df['v743b'].map(autonomy_map)
df['visiting_autonomy']    = df['v743d'].map(autonomy_map)

# v739 = 0 in our cleaned data means "no personal earnings" (our own sentinel,
# safe because DHS's real answer codes never use 0). Modeling choice: treat
# "no personal earnings" as low autonomy on this specific measure (score 0),
# since there's no income for her to have a say over.
df['earnings_autonomy'] = df['v739'].map(autonomy_map).fillna(0)
df.loc[df['v739'] == 0, 'earnings_autonomy'] = 0

# --- Simple recodes ---
df['is_urban']            = (df['v025'] == 1).astype(int)
df['is_working']          = df['v714']
df['has_bank_account']    = df['v170']
df['has_own_money_say']   = df['s929']
df['has_personal_earnings'] = (df['v739'] != 0).astype(int)
df['education_level']     = df['v106']   # already 0-3 ordinal
df['age_at_marriage']     = df['v511']

# --- Final feature set for PCA ---
feature_cols = [
    'caseid', 'v024', 'd005', 'sdweight',   # keep IDs/weight/state for later, not for PCA itself
    'education_level', 'is_urban', 'is_working', 'has_bank_account',
    'has_own_money_say', 'has_personal_earnings', 'age_at_marriage',
    'healthcare_autonomy', 'purchases_autonomy', 'visiting_autonomy', 'earnings_autonomy',
]

final = df[feature_cols]

print("Final feature table shape:", final.shape)
print()
print(final.head(10))
print()
print("Feature summary stats:")
print(final.describe())

final.to_csv("nfhs5_pca_ready.csv", index=False)
print("\nSaved nfhs5_pca_ready.csv -- ready for PCA.")
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

df = pd.read_csv("nfhs5_pca_ready.csv")

feature_cols = [
    'education_level', 'is_urban', 'is_working', 'has_bank_account',
    'has_own_money_say', 'has_personal_earnings', 'age_at_marriage',
    'healthcare_autonomy', 'purchases_autonomy', 'visiting_autonomy', 'earnings_autonomy',
]

X = df[feature_cols]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA()
pca.fit(X_scaled)

print("How much variance (information) each component explains:")
for i, var in enumerate(pca.explained_variance_ratio_, start=1):
    print(f"  PC{i}: {var*100:.1f}%")

print()
cumulative = pca.explained_variance_ratio_.cumsum()
print("Cumulative variance explained:")
for i, cum in enumerate(cumulative, start=1):
    print(f"  First {i} components: {cum*100:.1f}%")

print()
print("What each component is made of (loadings) -- first 3 components:")
loadings = pd.DataFrame(
    pca.components_[:3].T,
    index=feature_cols,
    columns=['PC1', 'PC2', 'PC3']
)
print(loadings.round(2))
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

df = pd.read_csv("nfhs5_pca_ready.csv")

feature_cols = [
    'education_level', 'is_urban', 'is_working', 'has_bank_account',
    'has_own_money_say', 'has_personal_earnings', 'age_at_marriage',
    'healthcare_autonomy', 'purchases_autonomy', 'visiting_autonomy', 'earnings_autonomy',
]
X = df[feature_cols]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=5)
X_pca = pca.fit_transform(X_scaled)

print("Testing different numbers of clusters (k):\n")
results = []
for k in range(2, 8):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_pca)
    inertia = km.inertia_
    sil = silhouette_score(X_pca, labels, sample_size=10000, random_state=42)
    results.append((k, inertia, sil))
    print(f"  k={k}:  inertia={inertia:,.0f}   silhouette={sil:.3f}")

chosen_k = 3

pd.set_option('display.width', 200)
pd.set_option('display.max_columns', 20)

km_final = KMeans(n_clusters=chosen_k, random_state=42, n_init=10)
df['cluster'] = km_final.fit_predict(X_pca)

print(f"\nFinal clustering with k={chosen_k}")
print("\nCluster sizes:")
print(df['cluster'].value_counts().sort_index())

print("\nAverage feature values per cluster (this is how we'll name/interpret them):")
print(df.groupby('cluster')[feature_cols].mean().round(2))

df.to_csv("nfhs5_clustered.csv", index=False)
print("\nSaved nfhs5_clustered.csv")
```

```python
import pandas as pd

# Pull just the district variable (plus caseid and state, to build a
# combined state+district key -- district codes restart at 1 within
# each state, so district alone isn't a unique identifier nationally)
reader = pd.read_stata(
    "IAIR7EFL.DTA",
    columns=["caseid", "v024", "sdist"],
    chunksize=50000,
    convert_categoricals=False,
)
district_df = pd.concat([c for c in reader], ignore_index=True)
district_df.to_csv("district_lookup.csv", index=False)
print("Saved district_lookup.csv, shape:", district_df.shape)

# Now merge onto our already-clustered data to check real sample sizes per district
clustered = pd.read_csv("nfhs5_clustered.csv")
merged = clustered.merge(district_df[["caseid", "sdist"]], on="caseid", how="left")

# Build a combined state+district key (since sdist restarts per state)
merged["state_district"] = merged["v024"].astype(str) + "_" + merged["sdist"].astype(str)

district_counts = merged.groupby("state_district").size().sort_values()
print("\nHow many districts total:", merged["state_district"].nunique())
print("\nSmallest district sample sizes:")
print(district_counts.head(15))
print("\nLargest district sample sizes:")
print(district_counts.tail(10))
print("\nHow many districts have fewer than 30 women (unreliable for %):")
print((district_counts < 30).sum(), "out of", len(district_counts))

merged.to_csv("nfhs5_clustered_with_district.csv", index=False)
print("\nSaved nfhs5_clustered_with_district.csv")
```

```python
import re
import json

names = {}
with open('IAIR7EFL.DO', encoding='utf-8', errors='replace') as f:
    lines = f.readlines()

flag = False
for line in lines:
    if line.strip().startswith('label define SDIST'):
        flag = True
        continue
    if flag and line.strip().startswith(';'):
        flag = False
    if flag:
        line = line.strip()
        m = re.match(r'^(\d+)\s+"(.+)"$', line)
        if m:
            code = int(m.group(1))
            name = m.group(2)
            names[code] = name

with open('sdist_code_to_name.json', 'w') as f:
    json.dump(names, f, indent=2)
```

```python
import pandas as pd
import json

df = pd.read_csv("nfhs5_clustered_with_district.csv")
df = df.dropna(subset=["sdist"])
df["sdist"] = df["sdist"].astype(int)

with open("sdist_code_to_name.json") as f:
    code_to_name = {int(k): v for k, v in json.load(f).items()}

state_names = {
    1: "Jammu & Kashmir", 2: "Himachal Pradesh", 3: "Punjab", 4: "Chandigarh",
    5: "Uttarakhand", 6: "Haryana", 7: "NCT of Delhi", 8: "Rajasthan",
    9: "Uttar Pradesh", 10: "Bihar", 11: "Sikkim", 12: "Arunachal Pradesh",
    13: "Nagaland", 14: "Manipur", 15: "Mizoram", 16: "Tripura",
    17: "Meghalaya", 18: "Assam", 19: "West Bengal", 20: "Jharkhand",
    21: "Odisha", 22: "Chhattisgarh", 23: "Madhya Pradesh", 24: "Gujarat",
    25: "Dadra & Nagar Haveli and Daman & Diu", 27: "Maharashtra",
    28: "Andhra Pradesh", 29: "Karnataka", 30: "Goa", 31: "Lakshadweep",
    32: "Kerala", 33: "Tamil Nadu", 34: "Puducherry",
    35: "Andaman & Nicobar Islands", 36: "Telangana", 37: "Ladakh",
}

df["district_name"] = df["sdist"].map(code_to_name)
df["state_name"] = df["v024"].map(state_names)

cluster_names = {
    0: "Homemaker w/ Household Voice",
    1: "Economically Active & Autonomous",
    2: "Fully Constrained",
}
df["archetype"] = df["cluster"].map(cluster_names)

# Group by the unique numeric code (not name) since 7 district names
# repeat across different states (e.g. "Bilaspur" exists in both
# Chhattisgarh and Himachal Pradesh)
weighted_sums = pd.pivot_table(
    df, values="d005", index="sdist", columns="archetype",
    aggfunc="sum", fill_value=0
)
district_breakdown = weighted_sums.div(weighted_sums.sum(axis=1), axis=0) * 100
district_breakdown = district_breakdown.round(1)
district_breakdown["n_women"] = df.groupby("sdist").size()

lookup = df.drop_duplicates("sdist").set_index("sdist")[["district_name", "state_name"]]
district_breakdown = district_breakdown.join(lookup).reset_index()

district_breakdown.to_csv("district_archetype_breakdown.csv", index=False)
```

```python
import pandas as pd
import geopandas as gpd

geo = gpd.read_file("india_states.geojson")
data = pd.read_csv("district_archetype_breakdown.csv")

name_fix_state = {
    "Andaman & Nicobar Islands": "Andaman and Nicobar Islands",
    "Dadra & Nagar Haveli and Daman & Diu": "Dadra and Nagar Haveli and Daman and Diu",
    "NCT of Delhi": "Delhi",
    "Jammu & Kashmir": "Jammu and Kashmir",
}
data["state_name_geo"] = data["state_name"].replace(name_fix_state)

manual_fix = {
    "Leh(Ladakh)": "Leh",
    "Muktsar": "Sri Muktsar Sahib",
    "Mewat": "Nuh",
    "Jyotiba Phule Nagar": "Amroha",
    "Kheri": "Lakhimpur Kheri",
    "Sant Ravidas Nagar (Bhadohi)": "Bhadohi",
    "Hugli": "Hooghly",
    "Baleshwar": "Balasore",
    "Dantewada": "Dakshin Bastar Dantewada",
    "Khargone (West Nimar)": "Khargone",
    "Khandwa (East Nimar)": "Khandwa",
    "Kachchh": "Kutch",
    "Bid": "Beed",
    "Sri Potti Sriramulu Nellore": "S.P.S. Nellore",
    "Gulbarga": "Kalaburagi",
    "North  District": "North Sikkim",
    "West District": "West Sikkim",
    "East District": "East Sikkim",
    "South District": "South Sikkim",
    "Mumbai Suburban": "Mumbai",
}
data["district_fixed"] = data["district_name"].replace(manual_fix)

archetype_cols = ["Economically Active & Autonomous", "Fully Constrained", "Homemaker w/ Household Voice"]

# Delhi: the map file only has ONE polygon for the whole state, so combine
# all Delhi sub-district rows into one, weighted by sample size
delhi_mask = data["state_name_geo"] == "Delhi"
delhi_rows = data[delhi_mask]
delhi_weighted = (delhi_rows[archetype_cols].multiply(delhi_rows["n_women"], axis=0)).sum() / delhi_rows["n_women"].sum()
delhi_combined = pd.DataFrame([{
    "district_fixed": "Delhi", "state_name_geo": "Delhi",
    "state_name": "NCT of Delhi", "n_women": delhi_rows["n_women"].sum(),
    **delhi_weighted.to_dict()
}])

# Mumbai + Mumbai Suburban: combine into one
mumbai_mask = (data["state_name_geo"] == "Maharashtra") & (data["district_fixed"] == "Mumbai")
mumbai_rows = data[mumbai_mask]
mumbai_weighted = (mumbai_rows[archetype_cols].multiply(mumbai_rows["n_women"], axis=0)).sum() / mumbai_rows["n_women"].sum()
mumbai_combined = pd.DataFrame([{
    "district_fixed": "Mumbai", "state_name_geo": "Maharashtra",
    "state_name": "Maharashtra", "n_women": mumbai_rows["n_women"].sum(),
    **mumbai_weighted.to_dict()
}])

final = data[~delhi_mask & ~mumbai_mask].copy()
final = pd.concat([final, delhi_combined, mumbai_combined], ignore_index=True)

final.to_csv("district_breakdown_final.csv", index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("nfhs5_clustered.csv")
cluster_names = {0: "Homemaker w/\nHousehold Voice", 1: "Economically Active\n& Autonomous", 2: "Fully\nConstrained"}
df["archetype"] = df["cluster"].map(cluster_names)

total_weight = df["d005"].sum()
national_pct = (df.groupby("archetype")["d005"].sum() / total_weight * 100)
national_pct = national_pct.reindex(["Homemaker w/\nHousehold Voice", "Economically Active\n& Autonomous", "Fully\nConstrained"])

fig, ax = plt.subplots(figsize=(8, 6))
bars = ax.bar(national_pct.index, national_pct.values, color=["#4C72B0", "#55A868", "#C44E52"])
for bar, val in zip(bars, national_pct.values):
    ax.text(bar.get_x() + bar.get_width()/2, val + 1, f"{val:.1f}%", ha="center", fontsize=12, fontweight="bold")
ax.set_ylabel("% of Women", fontsize=12)
ax.set_title("National Distribution of Women's Empowerment Archetypes\n(NFHS-5, k=3 Clustering, n=60,480)", fontsize=13)
ax.set_ylim(0, 70)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.tight_layout()
plt.savefig("national_cluster_chart.png", dpi=150)
print("Saved national_cluster_chart.png")
```
